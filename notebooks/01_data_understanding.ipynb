## 9. Basic Summary Statistics

This section covers basic data-quality checks: descriptive statistics, missing values, duplicate rows, and the number of unique values per column. We are only **observing** these facts — no values are removed or changed.

```python
# Descriptive statistics for all columns (numeric and categorical)
# include="all" so text/categorical columns are summarized too, not just numeric ones
df.describe(include="all")
```

```python
# Count missing values per column
missing_counts = df.isnull().sum()

# Keep only columns that actually have missing values, sorted from most to least
missing_summary = missing_counts[missing_counts > 0].sort_values(ascending=False)

# Also express missing values as a percentage of total rows
missing_percent = (missing_summary / len(df) * 100).round(2)

missing_summary_table = pd.DataFrame({
    "missing_count": missing_summary,
    "missing_percent": missing_percent
})

print(f"Columns with at least one missing value: {len(missing_summary_table)} out of {df.shape[1]}")
missing_summary_table
```

```python
# Count fully duplicated rows (every column identical)
n_duplicates = df.duplicated().sum()

print(f"Number of duplicated rows: {n_duplicates}")
```

```python
# Number of unique values for each column
# Useful for spotting identifier columns (very high uniqueness) or
# near-constant columns (very low uniqueness)
df.nunique().sort_values(ascending=False)
```
